```python
#recurssion : a function calling itself

#wap to find factorial of a number using recurssion

n=int(input("enter the number: "))

def fac(n):
  if n==1:
    return 1
  else:
    return n*fac(n-1)

fac(n)
```

```python
#wap to find sum of first n natural number using recurssion

n=int(input("enter the number: "))

def sum(n):
  if n==1:
    return 1
  else:
    return n+sum(n-1)
sum(n)
```

```python
#create a class of circles and print area and perimeter using methods

class Circle():
  def get_data(self):
    self.r=int(input("enter the radious"))

  def display_data(self):
    print(self.r)

  def area (self):
    print(3.14*self.r*self.r)

  def perimeter(self):
    print(2*3.14*self.r)

c1=Circle()

c1.get_data()
c1.display_data()
c1.area()
c1.perimeter()


```
